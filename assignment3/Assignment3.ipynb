(c) Plot a normal random variable
<font size = "3">

To get full points:

- Set a seed using "np.random.seed()"
- Create the random variable with the following parameters
- mean = 233, standard deviation = 99, and size = 1314

- Create a histogram
- Label the axes

```python
np.random.seed(3)
mean = 233
std_dev = 99
size = 1314
var_rand = np.random.normal(mean, std_dev, size)
plt.figure(figsize=[8,6])
plt.hist(var_rand)
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.title("Histogram of var_rand")
plt.show()


```
